```python
feature_cols = [
    "in_txs_degree",
    "out_txs_degree",
    "degree_centrality",
    "pagerank"
]
```

```python
graphsage_df = master_df[
    master_df["class"] != 3
].copy()

graphsage_df["class"] = graphsage_df["class"].map({
    1: 1,  # Fraud
    2: 0   # Legitimate
})

graphsage_df[feature_cols] = (
    graphsage_df[feature_cols]
    .fillna(0)
)

print(graphsage_df.shape)
print(graphsage_df["class"].value_counts())
```

```python
node_mapping = {
    node: idx
    for idx, node in enumerate(graphsage_df["txId"])
}

print("Nodes:", len(node_mapping))
```

```python
filtered_edges = tx_edges[
    tx_edges["txId1"].isin(node_mapping)
    &
    tx_edges["txId2"].isin(node_mapping)
]

print(filtered_edges.shape)
```

```python
import torch

edge_index = torch.tensor(
    [
        [node_mapping[src] for src in filtered_edges["txId1"]],
        [node_mapping[dst] for dst in filtered_edges["txId2"]]
    ],
    dtype=torch.long
)

print(edge_index.shape)
```

```python
x = torch.tensor(
    graphsage_df[feature_cols].values,
    dtype=torch.float
)

print(x.shape)
```

```python
y = torch.tensor(
    graphsage_df["class"].values,
    dtype=torch.long
)

print(y.shape)
```

```python
from torch_geometric.data import Data

data = Data(
    x=x,
    edge_index=edge_index,
    y=y
)

print(data)
```

```python
print(data)
```

```python
print(x.shape)
print(y.shape)
print(edge_index.shape)
```

```python
from sklearn.model_selection import train_test_split
import numpy as np
import torch

indices = np.arange(len(y))

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.2,
    stratify=y.numpy(),
    random_state=42
)

data.train_mask = torch.zeros(len(y), dtype=torch.bool)
data.test_mask = torch.zeros(len(y), dtype=torch.bool)

data.train_mask[train_idx] = True
data.test_mask[test_idx] = True

print("Train:", data.train_mask.sum().item())
print("Test:", data.test_mask.sum().item())
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv

class GraphSAGE(torch.nn.Module):
    def __init__(self, in_channels):
        super().__init__()

        self.conv1 = SAGEConv(in_channels, 64)
        self.conv2 = SAGEConv(64, 2)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return x
```

```python
model = GraphSAGE(data.num_features)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.01
)

print(model)
```

```python
for epoch in range(50):

    model.train()

    optimizer.zero_grad()

    out = model(
        data.x,
        data.edge_index
    )

    loss = F.cross_entropy(
        out[data.train_mask],
        data.y[data.train_mask]
    )

    loss.backward()

    optimizer.step()

    if epoch % 5 == 0:
        print(
            f"Epoch {epoch} | Loss {loss.item():.4f}"
        )
```

```python
from sklearn.metrics import classification_report

model.eval()

pred = model(
    data.x,
    data.edge_index
).argmax(dim=1)

print(
    classification_report(
        data.y[data.test_mask].cpu(),
        pred[data.test_mask].cpu()
    )
)
```
